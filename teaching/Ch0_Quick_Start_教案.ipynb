⏱ **时间分配：10 分钟**

🎯 **本段目标**
- 理解 MLP（多层感知机）的结构：输入层 -> 隐藏层 -> 输出层
- 理解关键组件：Linear 全连接层、ReLU 激活函数、Sigmoid 输出
- 掌握参数量计算方法：权重矩阵 + 偏置

🗣 **讲课话术**

> 现在我们来搭建网络。大家看 Cell 10 的代码，用的是 `nn.Sequential`，就像搭积木一样把层叠在一起。
>
> 我在白板上画一下结构：
> - 输入层：2 个神经元（因为每个数据点就 2 个坐标 x1, x2）
> - 隐藏层 1：32 个神经元，后面接 ReLU
> - 隐藏层 2：16 个神经元，后面接 ReLU
> - 输出层：1 个神经元，后面接 Sigmoid
>
> **类比时间**：把每一层想象成一个「投票委员会」。第一层 32 个委员各自看输入数据，给出自己的意见；第二层 16 个委员汇总第一层的意见，再给出更高级的判断；最后 1 个「主席」做最终决定——输出一个 0 到 1 之间的概率。
>
> **ReLU 是什么？** `max(0, x)`——负数变成 0，正数保持不变。它的作用是引入**非线性**。如果没有它，多层 Linear 叠在一起还是线性变换，画不了弯线。
>
> **Sigmoid 是什么？** 把任意实数压缩到 0~1 之间，可以理解为「概率」。接近 1 就是 Class 1，接近 0 就是 Class 0。
>
> 现在看 Cell 10 的输出：总参数量是 **641**。这 641 个数字就是网络要学习的全部内容。
>
> Cell 11 展示了每层的参数细节。比如第一层 `0.weight` 形状是 `[32, 2]`——32 个神经元，每个看 2 个输入，所以是 32x2=64 个权重；再加 32 个偏置 bias，一共 96 个参数。
>
> Cell 12 的表格总结了：96 + 528 + 17 = **641**。训练的目标就是调整这 641 个数字，让网络能正确分类。

👀 **输出要点**
- Cell 10：模型结构打印（6 层 Sequential），总参数量 641
- Cell 11：各层参数详情
  - `0.weight`: [32, 2] = 64 参数
  - `0.bias`: [32] = 32 参数
  - `2.weight`: [16, 32] = 512 参数
  - `2.bias`: [16] = 16 参数
  - `4.weight`: [1, 16] = 16 参数
  - `4.bias`: [1] = 1 参数

❓ **预判问题**

Q: 为什么选 32 和 16 个神经元？能用别的数字吗？
A: 完全可以。这是超参数，需要实验调。32 和 16 对这个小数据集够用了。后面 Extra 练习让大家试不同的数量。

Q: 为什么最后用 Sigmoid 而不是 ReLU？
A: 因为我们要输出概率（0~1 之间的数）。ReLU 的输出范围是 [0, +inf)，不适合当概率。Sigmoid 天然输出 (0, 1)。

Q: 641 个参数多不多？
A: 对这个问题来说足够了。作为对比，GPT-3 有 1750 亿参数，我们这个模型只有它的五百亿分之一。但原理是一样的——都是调数字。

## 段落 3：训练循环 Forward -> Loss -> Backward -> Step（Cell 13–19）

---

📍 **运行 Cell 13–19**

⏱ **时间分配：13 分钟**（其中练习约 5 分钟）

🎯 **本段目标**
- 理解深度学习核心四步循环
- 理解 BCE Loss 和 Adam 优化器的直觉
- 动手补全训练循环代码（练习 Cell 18）
- 观察 Loss 下降曲线

🗣 **讲课话术**

> 这是今天最重要的部分！深度学习的训练就四步，记住就行：
>
> **第一步 Forward（前向传播）**：把数据丢进网络，得到预测结果。就像考试——学生（网络）拿到题目（数据），写出答案（预测）。
>
> **第二步 Loss（计算损失）**：把预测和正确答案对比，算出差距有多大。我们用的是 BCE Loss——Binary Cross Entropy。Cell 14 有公式，但直觉很简单：如果真实标签是 1，你预测越接近 1，loss 越小；反之亦然。
>
> **第三步 Backward（反向传播）**：计算每个参数对 loss 的贡献——也就是梯度。哪个参数调一调能让 loss 降得最多？这一步 PyTorch 自动帮你算，一行 `loss.backward()` 搞定。
>
> **第四步 Step（更新参数）**：根据梯度调整参数。我们用 Adam 优化器（Cell 15-16），它比最基础的梯度下降更聪明——会自动调整每个参数的学习步长。
>
> 注意 Cell 13 提到的**梯度消失和梯度爆炸**——在很深的网络里，梯度往前传可能越来越小（消失）或越来越大（爆炸）。解决方法包括 BatchNorm、残差连接、梯度裁剪等，后续章节会详细讲。

---

### 🏋️ 练习：补全训练循环（Cell 17–18）

> 好，现在轮到大家动手了！Cell 18 有 4 个 TODO，请根据注释提示补全代码。这是深度学习最核心的代码模式，一定要自己写一遍。

**Hint 节奏：**
- **0–2 分钟**：自己尝试，看 TODO 旁边的提示
- **2 分钟**：第一个提示——四步分别用 `model(X_tensor)`, `criterion(y_pred, y_tensor)`, `loss.backward()`, `optimizer.step()`
- **4 分钟**：公布完整答案

**答案：**
```python
y_pred = model(X_tensor)           # Forward
loss = criterion(y_pred, y_tensor) # Loss
losses.append(loss.item())         # 记录
loss.backward()                    # Backward
optimizer.step()                   # Step
```

**常见错误：**
1. 忘记 `optimizer.zero_grad()` -> 梯度会累加，训练结果混乱
2. 把 `loss.backward()` 和 `optimizer.step()` 顺序搞反 -> 先算梯度才能更新
3. `losses.append(loss)` 而不是 `losses.append(loss.item())` -> 会保留计算图，内存泄漏

**验证标准：**
- Epoch 50 时 Loss 约 0.1183，Accuracy 约 95.2%
- Epoch 100 时 Accuracy 达到 100%
- 最终 Loss 约 0.0005
- 训练完成耗时约 0.4 秒

❓ **预判问题**

Q: 为什么要 `optimizer.zero_grad()`？
A: PyTorch 默认梯度会累加（这在某些场景有用，比如梯度累积）。每轮训练前必须清零，否则梯度越积越大，训练会崩。

Q: 500 轮够吗？怎么知道该训练多少轮？
A: 看 Loss 曲线。如果 Loss 不再下降了（趋于平坦），就可以停了。这里 100 轮之后 Accuracy 就已经 100% 了。

Q: `loss.item()` 和 `loss` 有什么区别？
A: `loss` 是一个 Tensor，带计算图，会占用内存。`.item()` 取出纯 Python 数字，不占额外内存。记录历史 loss 时一定要用 `.item()`。

Q: Adam 的学习率 0.01 是怎么选的？
A: 经验值。Adam 的常用学习率范围是 1e-4 到 1e-2。后面 Extra 练习会让大家试不同的值。

>

>
> **关键 takeaway**：神经网络不是只能画弯线，它能画**任意形状**的边界——只要网络够宽够深，数据够多。

❓ **预判问题**

Q: 为什么同心圆网络用 64-32 而不是 32-16？
A: 同心圆需要学习一个封闭边界（圆形），比月亮形的开放曲线更复杂，需要更多神经元来表达。用 32-16 也能学，但可能需要更多轮或者边界不够光滑。

Q: 为什么同心圆训练了 1000 轮而月亮形只训练了 500 轮？
A: 主要是为了让 Loss 降得更低、边界更光滑。其实 200 轮就已经 100% 了。
